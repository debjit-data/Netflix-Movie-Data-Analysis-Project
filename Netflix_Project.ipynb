```python
#****Categorised Popularity****#
#Four labels can be there
#Popular, average, Below average, Not Popular
```

```python
df['Vote_Average'].value_counts()
```

```python
def Categorised_Col(df,Col, labels):

  edges=[df[Col].describe()['min'],
         df[Col].describe()['25%'],
         df[Col].describe()['50%'],
         df[Col].describe()['75%'],
         df[Col].describe()['max']]
  df[Col]=pd.cut(df[Col], edges, labels=labels, duplicates='drop')
  return df
```

```python
Labels=['Not_Popular','Below_Average','Average','Popular']
df=Categorised_Col(df,'Vote_Average',Labels)
```

```python
df.dropna(inplace=True)
```
